```python
import pandas as pd
```

# Inspect the data

```python
products = pd.read_csv("./data/product_info.csv")
reviews = pd.read_csv("./data/reviews_0-250.csv")

print("Products: ", products.shape)
print("Reviews: ", reviews.shape)
```

```python
display(products.head())
display(reviews.head())
```

```python
print("Products shape:", products.shape)
print("Reviews shape:", reviews.shape)

print("\nDuplicate product IDs:")
print(products["product_id"].duplicated().sum())

print("\nMissing values in products:")
display(products.isna().sum().sort_values(ascending=False))

print("\nMissing values in reviews:")
display(reviews.isna().sum().sort_values(ascending=False))
```

```python
# Find review product IDs that don't exist in products
missing_product_ids = set(reviews["product_id"]) - set(products["product_id"])

print("Number of unmatched product IDs:", len(missing_product_ids))

print("Number of reviews with unmatched product IDs:")
print(reviews["product_id"].isin(missing_product_ids).sum())
```

```python
print("Unique products in reviews:", reviews["product_id"].nunique())

print(
    "Percentage of products represented:",
    round(
        reviews["product_id"].nunique()
        / products["product_id"].nunique() * 100,
        2
    ),
    "%"
)
```

```python
reviews_sample = reviews.sample(
    n=10000,
    random_state=42
).copy()

print("Sample shape:", reviews_sample.shape)

print(
    "Unique products represented:",
    reviews_sample["product_id"].nunique()
)
```

```python
product_columns = [
    "product_id",
    "product_name",
    "brand_name",
    "price_usd",
    "rating",
    "loves_count",
    "primary_category",
    "secondary_category",
    "sephora_exclusive"
]

review_columns = [
    "product_id",
    "author_id",
    "rating",
    "is_recommended",
    "submission_time",
    "review_text",
    "skin_type"
]

products_clean = products[product_columns].copy()
reviews_clean = reviews_sample[review_columns].copy()

print("Products:", products_clean.shape)
print("Reviews:", reviews_clean.shape)
```

```python
print("Product data types:")
print(products_clean.dtypes)

print("\nReview data types:")
print(reviews_clean.dtypes)

print("\nPreview:")
display(products_clean.head())
display(reviews_clean.head())
```
